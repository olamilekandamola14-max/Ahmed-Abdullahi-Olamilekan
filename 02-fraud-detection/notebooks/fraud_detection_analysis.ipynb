```python
# False Negative
FN = ((df["isFraud"] == 1) & (df["isFlaggedFraud"] == 0)).sum()

print("False Negative:", FN)
```

```python
# True Negative
TN = ((df["isFraud"] == 0) & (df["isFlaggedFraud"] == 0)).sum()

print("True Negative:", TN)
```

```python
# Precision
precision = TP / (TP + FP)

print(f"Precision: {precision:.2%}")
```

```python
# Recall
recall = TP / (TP + FN)

print(f"Recall: {recall:.2%}")
```

```python
accuracy = (TP + TN) / len(df)

print(f"Accuracy: {accuracy:.2%}")
```

Building The Model

```python
import numpy as np
import zipfile
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    roc_auc_score
)

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

file_path = "../../data/PS_20174392719_1491204439457_log.csv"

df = pd.read_csv(file_path)

df.head()
```

```python
target = "isFraud"
```

```python
features = [
    "step",
    "type",
    "amount"
]

X = df[features]
y = df["isFraud"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
numeric_features = [
    "step",
    "amount"
]

categorical_features = [
    "type"
]
```

```python
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)
numeric_transformer
```

```python
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print(preprocessor)
```

```python
# Build the Logistic Regression model
class_weight="balanced"
logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            LogisticRegression(
                class_weight="balanced",
                max_iter=1000,
                random_state=42
            )
        )
    ]
)
logistic_model
```

```python
logistic_model.fit(
    X_train,
    y_train
)
```

```python
y_pred = logistic_model.predict(X_test)
y_pred
```

```python

y_probability = logistic_model.predict_proba(X_test)[:, 1]
y_probability
```

```python
accuracy = accuracy_score(
    y_test,
    y_pred
)

print(f"Accuracy: {accuracy:.2%}")
```

```python
precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

print(f"Precision: {precision:.2%}")
```

```python
recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

print(f"Recall: {recall:.2%}")
```

```python
f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

print(f"F1 Score: {f1:.4f}")
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Non-Fraud", "Fraud"],
        zero_division=0
    )
)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred
)

print(cm)
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt=",d",
    xticklabels=["Non-Fraud", "Fraud"],
    yticklabels=["Non-Fraud", "Fraud"]
)

plt.title("Fraud Detection Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
roc_auc = roc_auc_score(
    y_test,
    y_probability
)

print(f"ROC-AUC: {roc_auc:.4f}")
```

Build a Random Forest model

```python
random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=1000,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)
random_forest_model
```

```python
random_forest_model.fit(
    X_train,
    y_train
)
```

```python
rf_pred = random_forest_model.predict(X_test)
rf_pred
```

```python
rf_probability = random_forest_model.predict_proba(
    X_test
)[:, 1]
rf_probability
```

```python
rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)

rf_precision = precision_score(
    y_test,
    rf_pred,
    zero_division=0
)

rf_recall = recall_score(
    y_test,
    rf_pred,
    zero_division=0
)

rf_f1 = f1_score(
    y_test,
    rf_pred,
    zero_division=0
)

rf_auc = roc_auc_score(
    y_test,
    rf_probability
)
print(f"Accuracy : {rf_accuracy:.2%}")
print(f"Precision: {rf_precision:.2%}")
print(f"Recall   : {rf_recall:.2%}")
print(f"F1 Score : {rf_f1:.4f}")
print(f"ROC-AUC  : {rf_auc:.4f}")
```

```python
# Compare the Models
model_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest"
    ],
    
    "Accuracy": [
        accuracy,
        rf_accuracy
    ],
    
    "Precision": [
        precision,
        rf_precision
    ],
    
    "Recall": [
        recall,
        rf_recall
    ],
    
    "F1 Score": [
        f1,
        rf_f1
    ],
    
    "ROC-AUC": [
        roc_auc,
        rf_auc
    ]
})

model_comparison
```
